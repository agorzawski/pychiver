## Service init

```python
from pychiver.saveandrestore import SaveAndRestore
sar = SaveAndRestore(service_url="http://jmasar.tn.esss.lu.se")
```

## Search available snapshots 

> For now, there is no SaveRestore (MASAR) API for the string search. Therefore the only way to get all snapshots is to browse the root folder for all unique IDs

```python
print('=== Get Snapshots ')
test = sar.getSnapshots(configUniqueId='c2d7bf79-c6b9-4bb1-83be-c50a6ebc8776')
```

```python
for one in test.values():
    print('Snapshot name {} and uniqueID {}'.format(one.name, one.uniqueId))
    snapshotStoredValues = one.getStoredValues()
    snapshotComparisonResult = sar.compare(snapshot=one)
```

```python
snapshotStoredValues
```

```python
snapshotComparisonResult
```

```python
print('===== All Config Sets')
confs = sar.getConfigurations()
for k, c in confs.items():
    print(k, c)
```
