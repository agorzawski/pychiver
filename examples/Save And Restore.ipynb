```python
from pychiver.saveandrestore import SaveAndRestore
sar = SaveAndRestore(service_url="http://jmasar.tn.esss.lu.se")
```

```python
print('=== Get Snapshots ')
test = sar.getSnapshots(configUniqueId='c2d7bf79-c6b9-4bb1-83be-c50a6ebc8776')
```

```python
for one in test.values():
    storedValues = one.getStoredValues()
    result = sar.compare(snapshot=one)
```
